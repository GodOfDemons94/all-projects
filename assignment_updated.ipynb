```python
# Calculating Mutual Information Score of each variable to understand
# how strong each variable has an immpact on the diabetes:
from sklearn.feature_selection import mutual_info_classif

_X = df_X.astype('int64')
discrete_features = _X.dtypes == int

def make_mi_scores(X, y, discrete_features):
    mi_scores = mutual_info_classif(X, y, discrete_features=discrete_features)
    mi_scores = pd.Series(mi_scores, name="MI Scores", index=_X.columns)
    mi_scores = mi_scores.sort_values(ascending=False)
    return mi_scores

mi_scores = make_mi_scores(_X, df_Y, discrete_features)
mi_scores  # show a few features with their MI scores
```

```python
# plotting the MI scores on a horizontal bar chart for visualization:
import numpy as np
def plot_mi_scores(scores):
    scores = scores.sort_values(ascending=True)
    width = np.arange(len(scores))
    ticks = list(scores.index)
    plt.barh(width, scores)
    plt.yticks(width, ticks)
    plt.title("Mutual Information Scores")

plt.figure(dpi=100, figsize=(8, 5))
plot_mi_scores(mi_scores)
```

**From the above comparision of MI scores, it can be easily concluded that the values of GenHlth has the most impact on the target values whereas the 'NoDocbcCost' has the least impact. Besides this, the Age, BMI, HighBP and High Cholestrol can be seen impacting the target values significantly which seems very obvious.**

## 4. Model Building and Evaluation

**We will now check the cross validation accuracy of the following classification models on our dataset and select the model with the best accuracy to train the model.**

1. Random Forest Classifier
2. XGBoost
3. Logistic Regression
4. Support Vector Classifier (SVC)

```python
# evaluating Random Forest Classifier model:
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import cross_val_score
rf_model = RandomForestClassifier(random_state = 42)
cv = cross_val_score(rf_model, df_scaled, df_Y, cv=5)
print(cv)
print(cv.mean())
```

```python
# Evaluating XGBoost model:
import xgboost as xgb
xgb_model = xgb.XGBClassifier(random_state=42)
cv = cross_val_score(xgb_model, df_scaled, df_Y, cv=5)
print(cv)
print(cv.mean())
```

```python
# Evaluating Logistic Regression model:
from sklearn.linear_model import LogisticRegression
lr_model = LogisticRegression(max_iter = 1000)
cv = cross_val_score(lr_model, df_scaled, df_Y, cv=5)
print(cv)
print(cv.mean())
```

**From the above accuracy scores of different models on the crosss validation data, it can be seen that the XGBoost model is performing slightly better than the other classification models. Hence, we will train this model on our dataset.**

(Note: The SVC model has not been evaluated as it is taking very long time to run. It is adviced to evaluate this model to further improve the decision making on model selection)

```python
# Constructing pram_grid and setting the best parameters of our XGBoost model using GridSearchCV:
from sklearn.model_selection import GridSearchCV

param_grid = {
    'n_estimators': [100, 200],
    'learning_rate': [0.01, 0.1],
    'max_depth': [3, 4],
    'gamma': [0.1],
    'reg_alpha': [0, 0.01],
    'reg_lambda': [1, 0.1]
}
grid_search = GridSearchCV(estimator=xgb_model,param_grid=param_grid, cv=5, n_jobs=-1, verbose=2)
```

**Splitting our dataset into train data and test data. Training XGBoost model on train data and then predicting and evaluating the outcome of the model using test data.** 

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import f1_score

X_train, X_test, Y_train, Y_test = train_test_split(df_scaled, df_Y, test_size=0.2, random_state=42)
xgb_model = grid_search.fit(X_train, Y_train)
Y_hat = xgb_model.predict(X_test)
f1score = f1_score(Y_test, Y_hat, average='micro')
print('Test F1 Score (Micro):', f1score)
```

**It can be concluded here that the best f1 score on the test data that can be achieved while predicting whether a person can be diagnosed with diabetes is 84.24% with XGBoost Machine Learning model.**


We'll further check if we can improve this result using neural network.

```python
# This cell performs the following operation:
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout
from sklearn.model_selection import train_test_split
from tensorflow.keras.utils import to_categorical

# Set a random seed for reproducibility
tf.random.set_seed(42)

# Define the model architecture
model = Sequential([
    Dense(64, activation='relu'),  # Input layer
    Dropout(0.5), #Dropout layer
    Dense(32, activation='relu'), # Hidden layer 1
    Dropout(0.5), #Dropout layer
    Dense(3, activation='softmax') # Output layer
])

# One-hot encode the target variable
df_Y_encoded = to_categorical(df_Y, num_classes=3)

# Compile the model
model.compile(optimizer = tf.keras.optimizers.Adam(learning_rate=0.1),
              loss=tf.keras.losses.CategoricalCrossentropy(from_logits=True),
              metrics=['accuracy'])

#Splitting the data
X_train, X_test, Y_train, Y_test = train_test_split(df_scaled, df_Y_encoded, test_size=0.2, random_state=42)

# Assuming you have training data (X_train, y_train) and testing data (X_test, y_test)
model.fit(X_train, Y_train, epochs=25, batch_size=32, validation_split=0.2)

# Evaluate the model on the test data
Y_hat_probabilities = model.predict(X_test)
Y_hat_classes = np.argmax(Y_hat_probabilities, axis=1) # Convert probabilities to class labels
Y_test_classes = np.argmax(Y_test, axis=1) # Convert one-hot encoded test labels to class labels

f1score = f1_score(Y_test_classes, Y_hat_classes, average='micro')
print('Test F1 Score (Micro):', f1score)
```

**Hence, it is concluded that a simple neural network can not beat the f1 score outcome of XGBoost model.**
